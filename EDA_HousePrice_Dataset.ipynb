```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
fig,axes=plt.subplots(3,2,figsize=(15,15))
fig.tight_layout(pad=10)

sns.histplot(df['SalePrice'],kde=True,ax=axes[0,0],color='blue')
axes[0,0].set_title('Distribution of SalePrice')

sns.histplot(df['GrLivArea'],kde=True,ax=axes[0,1],color='green')
axes[0,1].set_title('Distribution of GrLivArea')

sns.countplot(x='OverallQual', data=df, ax=axes[1, 0], palette='viridis')
axes[1, 0].set_title('Distribution of OverallQual')

sns.histplot(df['YearBuilt'], kde=True, ax=axes[1, 1], color='orange')
axes[1, 1].set_title('Distribution of YearBuilt')

sns.histplot(df['TotalBsmtSF'], kde=True, ax=axes[2, 0], color='purple')
axes[2, 0].set_title('Distribution of TotalBsmtSF')

sns.histplot(df['LotArea'], kde=True, ax=axes[2, 1], color='red')
axes[2, 1].set_title('Distribution of LotArea')


plt.show()
```

```python
numeric_df=df.select_dtypes(include=[np.number])
corr=numeric_df.corr()

k=10
top_cols=corr.nlargest(k,'SalePrice')['SalePrice'].index
top_corr=df[top_cols].corr()

plt.figure(figsize=(10,8))
sns.heatmap(top_corr, annot=True, cmap='coolwarm', fmt=".2f", square=True, cbar_kws={"shrink": .8})
plt.title("Correlation Heatmap: Top 10 Numeric Features vs SalePrice")
plt.show()
```

```python
top_5_features = top_cols[1:6]
print(f"Top 5 correlated features: {list(top_5_features)}")
```

```python
fig, axes = plt.subplots(1, 5, figsize=(25, 5))
for i, feature in enumerate(top_5_features):
    sns.scatterplot(x=df[feature], y=df['SalePrice'], ax=axes[i], alpha=0.5, color='darkblue')
    axes[i].set_title(f'SalePrice vs {feature}\nCorr: {corr.loc[feature, "SalePrice"]:.2f}')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns


fig, axes = plt.subplots(2, 2, figsize=(15, 10))


sns.boxplot(x=df['SalePrice'], ax=axes[0, 0], color='skyblue')
axes[0, 0].set_title('Box Plot of SalePrice')

sns.boxplot(x=df['GrLivArea'], ax=axes[0, 1], color='lightgreen')
axes[0, 1].set_title('Box Plot of GrLivArea')

sns.boxplot(x=df['TotalBsmtSF'], ax=axes[1, 0], color='salmon')
axes[1, 0].set_title('Box Plot of TotalBsmtSF')

sns.boxplot(x=df['LotArea'], ax=axes[1, 1], color='plum')
axes[1, 1].set_title('Box Plot of LotArea')

plt.tight_layout()
plt.show()
```

```python
# Drop properties over 4,000 sq ft that sold for strangely low prices
df = df.drop(df[(df['GrLivArea'] > 4000) & (df['SalePrice'] < 300000)].index)
```

```python
sns.scatterplot(x=df['GrLivArea'], y=df['SalePrice'],alpha=0.5, color='darkblue')
```

```python
# Apply log(1+x) transform to normalize the target variable
df['SalePrice_Log'] = np.log1p(df['SalePrice'])
```

```python
sns.histplot(df['SalePrice_Log'],kde=True,color='blue')
```

```python
# Apply log(1+x) transform to compress extreme acreage
df['LotArea_Log'] = np.log1p(df['LotArea'])
```

```python
sns.histplot(df['LotArea_Log'], kde=True,color='red')
```

```python
# Multicollinearity (dropping columns)
df = df.drop(columns=['GarageCars', '1stFlrSF'])
```

```python
df['OverallQual_Sq'] = df['OverallQual'] ** 2
```

```python
print(f"New dataset shape: {df.shape}")
print("New columns created:", ['SalePrice_Log', 'LotArea_Log', 'OverallQual_Sq'])
```
